**DATA CLEANING**

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('zomato.csv',encoding='latin-1')
```

```python
df.isnull().sum()
```

```python
print(df.shape)
print(df.head())
print(df.tail())
print(df.info())
print(df.columns)
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["online_order"].unique()
```

```python
df.drop(columns=['Unnamed: 0.1','Unnamed: 0'],inplace=True)
df
```

```python
df["rate (out of 5)"].fillna(df["rate (out of 5)"].mean(),inplace=True)
df
```

```python
df["avg cost (two people)"].fillna(df["avg cost (two people)"].median(),inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
for i in ["restaurant name","restaurant type","cuisines type","area","local address"]:
    df[i]=df[i].str.strip()
```

```python
df.duplicated().sum()
```

```python
df.to_csv("zomato_cleaned.csv",index=False)
```
