Task 4.3: Pandas Data Analysis
Create a Pandas DataFrame with 10 students and the following columns: Name, Maths, Science, English. (Fill with random marks between 50 and 100).

```python
dataf=pd.DataFrame({"Name":["ankit","aman","harkesh","pushkal","charchil","vikas","krish","harsh vardhan","ansh negi","rakshit"],
"Maths":np.random.randint(50,101,10),
"Science":np.random.randint(50,101,10),
"English":np.random.randint(50,101,10)})
```

Perform the following operations:
Print the average marks of each subject

```python
dataf[["Maths","Science","English"]].mean()
```

Add a new column Total which is the sum of marks across the three subjects

```python
dataf["Total"]=dataf[["Maths","Science","English"]].sum(axis=1)
dataf
```

⦁	Print the student with the highest total marks

```python
dataf.loc[dataf["Total"].idxmax()]
```

⦁Add another column Result where the value is 'Pass' if the student has total marks ≥ 150, otherwise 'Fail

```python
dataf["RESULT"]=np.where(dataf["Total"]>=150,"PASS","FAIL")
```

```python
dataf
```

⦁	Sort the DataFrame by Total in descending order


```python
dataf=dataf.sort_values("Total",ascending=False)
dataf
```
