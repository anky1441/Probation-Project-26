Task 3.1: NumPy Basics

1.	Create a 1D NumPy array of 10 random integers between 1 and 100.


```python
import numpy as np
arr1=np.random.randint(1,101,size=10)
arr1
```

```python
print(np.max(arr1))
```

⦁	Print the minimum element of the array.

```python
print(np.min(arr1))
```

⦁	Print the mean of the array.

```python
print(np.mean(arr1))
```

⦁	Print the sum of all elements.


```python
print(np.sum(arr1))
```

⦁	Find and print the index of the maximum element.

```python
print(np.argmax(arr1))
```

⦁	Find and print the index of the minimum element.


```python
print(np.argmin(arr1))
```

⦁	Sort the array in ascending order and print it.

```python
print(np.sort(arr1))
```
